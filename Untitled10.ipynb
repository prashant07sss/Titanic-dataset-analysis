Q12. Difference Between View and Copy

```python


a = np.array([1, 2, 3, 4, 5])

# View
view = a[1:4]
view[0] = 100

print("Original after view:", a)

# Copy
copy = a[1:4].copy()
copy[0] = 200

print("Original after copy:", a)
print("Copy:", copy)
```

Q39. Student Marks Analysis

```python
marks = np.array([
    [60,70,80],
    [40,50,45],
    [80,90,85],
    [30,40,35],
    [55,60,65],
    [70,75,80],
    [45,50,55],
    [90,85,95],
    [50,55,60],
    [35,45,40]
])

total = np.sum(marks, axis=1)
average = np.mean(marks, axis=1)

print("Total:", total)
print("Average:", average)
print("Students with average >= 50:", np.where(average >= 50)[0])
```
